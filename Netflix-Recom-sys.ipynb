# Text preprocessing 

```python
df['tags']=df["overview"].fillna(' ')+df["keywords"].fillna(" ")+df["genres"].fillna(" ")
df["tags"]
```

```python
df.isna().sum()
```

```python
from nltk.stem import PorterStemmer
```

```python
ps= PorterStemmer()
```

```python
def preprocess(text):
    lower = text.lower()
    y =[]
    for i in lower.split():
        y.append(ps.stem(i))
    return ' '.join(y)
```

```python
df["tags"]=df["tags"].apply(preprocess)
df["tags"]
```

# Text to vectors 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf=TfidfVectorizer(max_features=15000,stop_words="english")
```

```python
vectors=tf.fit_transform(df["tags"])
vectors
```

```python
tf.get_feature_names_out().tolist()
```

# cosine similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
similarity
```

```python
similaritydf=pd.DataFrame(similarity,index=df["title"],columns=df["title"])
```

```python
similaritydf["Avatar"].sort_values(ascending=False)[1:5]
```

```python
similaritydf["Interstellar"].sort_values(ascending=False)[1:5]
```

```python
df[df["title"]=="Avatar"].index[0]
```

```python
def recommend(movie):
    idx = df[df['title'] == movie].index[0]
    movie_list = np.argsort(similarity[idx])[::-1][1:6]
    return df.iloc[movie_list].title.tolist()
```

```python
recommend('The Dark Knight')
```
